```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
# --- 2. Prepare the Text ---
# Combine all job descriptions into a single string.
# We'll also drop any empty rows to avoid errors.
text = " ".join(desc for desc in df.description.dropna())

print(f"Total words in the combined text: {len(text.split())}")


# --- 3. Customize Stopwords ---
# Create a set of words to ignore. We start with the default STOPWORDS list
# and add custom words that are common in job descriptions but not insightful.
# For a "Data Engineer" search, 'data' and 'engineer' will be dominant, so we can remove them.
stopwords = set(STOPWORDS)
stopwords.update(["data", "engineer", "engineering", "experience", "work", "team", "job", "required", "skills", "role"])


# --- 4. Generate the Word Cloud ---
# Create the WordCloud object with desired settings (background, size, stopwords).
wordcloud = WordCloud(
    stopwords=stopwords,
    background_color="white",
    width=800,
    height=400,
    colormap='viridis', # A visually appealing color map
    max_words=100,      # Limit the number of words shown
    contour_width=1,
    contour_color='steelblue'
).generate(text)


# --- 5. Display and Save the Image ---
print("\nGenerating word cloud image...")

# Use matplotlib to display the generated image.
plt.figure(figsize=(12, 6)) # Set the figure size
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off") # Hide the axes
plt.show()

# Save the generated image to a file.
wordcloud.to_file("data_engineer_wordcloud.png")

print("\nWord cloud saved as 'data_engineer_wordcloud.png'")
```
